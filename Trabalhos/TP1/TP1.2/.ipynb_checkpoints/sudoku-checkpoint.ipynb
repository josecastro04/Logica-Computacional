```python
from ortools.sat.python import cp_model

class Sudoku_Solver:
    def __init__(self, n):
        self.n = n
        self.tamaho = n * n
        self.model = cp_model.CpModel()
        self.matriz = [[self.model.NewIntVar(1, self.N, f"m_{l}_{c})"
                                            for c in range(self.N)]
                       for l in range(self.N)
                      ]
    def add(self, l, c, val=None)
        if not (1 <= l <= tamanho and 1 <= c <= tamanho):
            raise Exception("Tem que indicar uma célula dentro dos limites")
```

```python
from z3 import Solver, Distinct, sat, Int

tamanho = n**n

matriz = []
for r in range(tamanho):
    l = []
    for c in range(n):
        l.append(Int("m_%s_%s" % (r + 1, c + 1)))
    matriz.append(l)

print(matriz[0])

##verificar linhas
linhas = []
for r in range(tamanho):
        linhas.append(Distinct(matriz[r]))

##verificar colunas
colunas = []
for c in range(tamanho):
    colunas.append(Distinct(matriz[r][c] for r in range(n)))

##verificar quadrados
quadrados = []
for q_l in range(n):
    for q_c in range(n):
        quadrados.append(Distinct(matriz[q_l * n + r][q_c * n + c]) for r in range(n) for c in range(n))
    
```
